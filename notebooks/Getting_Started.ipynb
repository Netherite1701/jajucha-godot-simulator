## Read a camera image

Each camera image is a 640 × 400 BGR NumPy array. This first call waits for a fresh simulator frame.


```python
from jchm import camera
import cv2

image = camera.get_image('center')
print(image.shape, image.dtype)
camera.show_image(image)
```

## LiDAR and motor units

LiDAR angles are degrees, ranges are millimetres. The motor API uses the same controller convention as the supplied Jajucha script. Start at low speed and keep Emergency stop (F8) available.


```python
from jchm import lidar, control

angles_deg, ranges_mm = lidar.get_lidar()
print(len(angles_deg), 'rays')
lidar.show_lidar(angles_deg, ranges_mm)

# Example low-speed command: left servo, right servo, motor
# control.set_motor(0, 0, 2)
```
